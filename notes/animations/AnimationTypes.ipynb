# Transform animations

## Examples

### Transform animations

```python
class TransformAnimations(Scene):
    '''Using Transfrom and Mobject.copy()
    Circunference concept'''
    def construct(self):
        t = r'\theta'
        m = 2
        x = MathTex(r'x = ',
                    'r',r'\cos',
                    '(',r'\theta',
                    ')').move_to(np.array([m,m,0]))
        y = MathTex(r'y = ',
                    'r',r'\sin',
                    '(',r'\theta',
                    ')').move_to(np.array([-m,m,0]))
        r = MathTex(f'r','^2 = ',
                    r'\cos',f'^2(',
                    f'{t}',') + ',
                    r'\sin','^2(',
                    f'{t}',')').move_to(np.array([0,0.75,0]))
        self._Write(x)
        self._Write(y)
        self.play(Transform(x[1].copy(),r[0]),
                  Transform(x[2].copy(),r[2]),
                  Transform(x[-2].copy(),r[4]),
                  Transform(y[1].copy(),r[0]),
                  Transform(y[2].copy(),r[6]),
                  Transform(y[-2].copy(),r[8]), run_time=3)
        self.play(Write(r[1]),
                  Write(r[3]),
                  Write(r[5]),
                  Write(r[7]),
                  Write(r[9]), run_time = 2)
        self.wait()

        c = Circle(radius=1.5, fill_color='#614FE8',
                   stroke_color='#46E838',
                   fill_opacity = 0.75).move_to(np.array([0,-1.75,0]))
        self.play(Create(c))

    def _moveCopy(self, a : Mobject, b : Mobject, wt = 0.5):
        self.play(Transform(a.copy(),b))
        self.wait(wt)

    def _Write(self, obj: Mobject , rt = 1 , wt = 0.5):
        self.play(Write(obj), run_time=rt)
        self.wait(wt)
```

# .animate syntax

From

    Mobject.animate

```python
# @author : Alexis Tercero
class AnimateSyntax(Scene):
    '''Animation by Mobjecte.animate syntax.'''
    def construct(self):
        # Mobjects
        txt = Tex('Alexis').to_edge(UL, buff=0.5)
        sq = Square(side_length=0.5, 
                    fill_color=GREEN, 
                    fill_opacity=0.75)
        tri = Triangle().scale(0.6).to_edge(DR)

        # Creation animations
        self.play(Write(txt)) 
        self.play(DrawBorderThenFill(sq), run_time=2) 
        self.play(Create(tri)) 
        self.wait(2)

        # .animate syntax for motion animation
        rt = 2
        self.play(txt.animate.to_edge(UR),
                  run_time=rt)
        self.play(sq.animate.to_edge(LEFT),
                  run_time=rt)
        self.play(sq.animate.move_to(ORIGIN),
                  run_time=rt)
        self.play(tri.animate.to_edge(DL),
                  run_time=rt)
        self.play(tri.animate.to_edge(DR),
                  txt.animate.to_edge(UL),
                  sq.animate.scale(2),
                  run_time=rt)
        self.wait()
```

# Creation Mobject animation types

## Examples

### Write, Create, DrawBorderThenFill

```python
# @author : Alexis Tercero
class AnimationTest(Scene):
    def construct(self):
        l = 3
        txt = Tex('Alexis')

        sq1 = Square(side_length=l,
                    stroke_color = '#3E78FF',
                    fill_color = '#B38007',
                    fill_opacity = 0.75
                    )
        sq1.set_opacity(0)

        sq2 = Square(side_length=l,
                    stroke_color = '#EB4B23',
                    fill_color = '#19B377',
                    fill_opacity = 0.75
                    )
        sq2.set_opacity(0)

        tri = Triangle()
        tri.set_opacity(0)

        # creatw and arrange VGroup
        G = VGroup(txt,sq1,sq2,tri)
        G.arrange_in_grid(rows=2, buff=0.1)
        self.add(G)

        # Showing some creation mobjects animation types
        self.play(Write(txt), run_time=2)
        
        sq1.set_opacity(1)
        self.play(Create(sq1), run_time=2)

        sq2.set_opacity(1)
        self.play(DrawBorderThenFill(sq2), run_time=2)

        tri.set_stroke(opacity=1)
        self.play(Create(tri), run_time=2)
        
        # self.play(ShowSubmobjectsOneByOne(G,run_time=3))
```
