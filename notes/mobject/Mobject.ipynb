# Mobject.arrange()

## Intro

Sorts Mobject next to each other on screen

    arrange(direction=array([1., 0., 0.]), 
            buff=0.25, 
            center=True, 
            **kwargs)

## Examples

```python
class Example(Scene):
    def construct(self):
        l = 0.8
        s1 = Square(l)
        s2 = Square(l)
        s3 = Square(l)
        s4 = Square(l)
        x = VGroup(s1, s2, s3, s4).set_x(0).arrange(
            direction=np.array([1., 1., 0.]),
            buff=0.2
            )
        self.add(x)
```

# Mobject.arrange_in_grid()

## Intro

Arrange submobjects in a grid.

    arrange_in_grid(rows=None, 
                    cols=None, 
                    buff=0.25, 
                    cell_alignment=array([0., 0., 0.]), 
                    row_alignments=None, 
                    col_alignments=None, 
                    row_heights=None, 
                    col_widths=None, 
                    flow_order='rd', 
                    **kwargs)

```python
'''
Usually
'''
Mobject.arrange_in_grid()
'''
takes a vectorized group of Mobjects (VGroup) as input.
'''
```

## Examples

### ExampleBoxes

```python
class ExampleBoxes(Scene):
    def construct(self):
        boxes=VGroup(*[Square() for s in range(0,6)])
        boxes.arrange_in_grid(rows=2, buff=0.1)
        self.add(boxes)
```

### ArrangeInGrid

```python
class ArrangeInGrid(Scene):
    def construct(self):
        boxes = VGroup(*[
            Rectangle(WHITE, 0.5, 0.5).add(Text(str(i+1)).scale(0.5))
            for i in range(24)
        ])
        self.add(boxes)

        boxes.arrange_in_grid(
            buff=(0.25,0.5),
            col_alignments="lccccr",
            row_alignments="uccd",
            col_widths=[1, *[None]*4, 1],
            row_heights=[1, None, None, 1],
            flow_order="dr"
        )
```
